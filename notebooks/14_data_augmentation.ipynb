```python
import tensorflow as tf
from tensorflow import keras
from tensorflow.keras import layers
from tensorflow.keras.models import Sequential
from keras.preprocessing import image
from tensorflow.keras.preprocessing.image import ImageDataGenerator
```

```python
img=image.load_img('/content/readhead-cat-sitting.jpg',target_size=(350,350))
```

```python
import matplotlib.pyplot as plt
plt.imshow(img)
```

```python
from textwrap import fill
datagen=ImageDataGenerator(
    rotation_range=30,
    width_shift_range=0.2,
    height_shift_range=0.2,
    horizontal_flip=True,
    zoom_range=0.2,
    fill_mode='nearest',
    shear_range=0.2)
```

```python
img=image.img_to_array(img)
```

```python
img.shape
```

```python
input_batch=img.reshape(1,350,350,3)
```

```python
i=0
for output in datagen.flow(input_batch,batch_size=1):
  plt.figure(i)
  plot=plt.imshow(image.array_to_img(output[0]))
  i+=1
  if i==10:
    break
```
